# Final Report Analysis Notebook
Compares No-DP, LDP, GDP, and LDP+GDP setups.


```
import numpy as np
import os
import matplotlib.pyplot as plt
```

## Load Experiment Results

```
setups = ['no_dp','ldp','gdp','both']
base_path = 'results'
results = {}
for s in setups:
    path = os.path.join(base_path, s, 'server_output', 'global_model.npz')
    if os.path.exists(path):
        data = np.load(path)
        results[s] = {
            'accuracy': float(data['avg_client_accuracy']),
            'samples': int(data['total_samples'])
        }
print(results)
```

## Accuracy Comparison

```
labels = list(results.keys())
acc = [results[k]['accuracy'] for k in labels]
plt.figure()
plt.bar(labels, acc)
plt.title('Accuracy Across DP Setups')
plt.xlabel('Setup')
plt.ylabel('Accuracy')
plt.show()
```

## Noise vs Accuracy Trend (Manual Input)

```
# Fill manually after experiments
noise = [0, 0.05, 0.1, 0.5]
accuracy = [0,0,0,0]
plt.figure()
plt.plot(noise, accuracy, marker='o')
plt.xlabel('Noise Scale')
plt.ylabel('Accuracy')
plt.title('Privacy vs Accuracy')
plt.show()
```

## Fairness Trend (DP Gap)

```
dp_gap = [0,0,0,0]
plt.figure()
plt.plot(noise, dp_gap, marker='o')
plt.xlabel('Noise Scale')
plt.ylabel('DP Gap')
plt.title('Privacy vs Fairness')
plt.show()
```
